```python
p_IMU_global = np.array([0, -9, 0, 0])
p_face_global = np.array([0, -9, 5, 0])
# p_face_cam = np.array([0, 0, 0, 5])

q_IMU_global = np.array([np.cos(np.deg2rad(6)), np.sin(np.deg2rad(6)), 0, 0])

scene_IMU_translation = np.array([0, 0, -0.0013, -0.00662])
scene_IMU_rotation = np.array([np.cos(np.deg2rad(-51)), np.sin(np.deg2rad(-51)), 0, 0])

q_IMU_global
```

## helpers

```python
def left_quat_matrix(q):
    qw, qx, qy, qz = q[0], q[1], q[2], q[3]
    
                            #w
                            #x
                            #y
                            #z
    return np.array([
        [qw, -qx, -qy, -qz], 
        [qx,  qw, -qz,  qy], 
        [qy,  qz,  qw, -qx], 
        [qz, -qy,  qx,  qw]  
    ])

def quat_inv(q):
    return np.array([q[0], -q[1], -q[2], -q[3]])
```

# from global to IMU

```python
translated = p_face_global - p_IMU_global
first_rotation = left_quat_matrix(quat_inv(q_IMU_global)) @ translated
second_rotation = left_quat_matrix(first_rotation) @ q_IMU_global

p_face_IMU = second_rotation

p_face_IMU
```

# from global to face

```python
translated2 = p_face_IMU - scene_IMU_translation
first_rotation2 = left_quat_matrix(quat_inv(scene_IMU_rotation)) @ translated2
second_rotation2 = left_quat_matrix(first_rotation2) @ scene_IMU_rotation

p_face_cam = second_rotation2

p_face_cam
```
